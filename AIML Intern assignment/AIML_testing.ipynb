```python
import numpy as np
from attention import scaled_dot_product_attention
from utils import create_mask

# Example Q, K, V
Q = np.array([[[1, 0], [0, 1]]], dtype=np.float32)
K = np.array([[[1, 0], [0, 1]]], dtype=np.float32)
V = np.array([[[1, 2], [3, 4]]], dtype=np.float32)

# Without mask
out, weights = scaled_dot_product_attention(Q, K, V, mask=None)
print("Attention Output:\n", out)
print("Attention Weights:\n", weights)

# Causal mask manually (since create_mask doesn't have 'causal')
def causal_mask(size):
    mask = np.tril(np.ones((size, size), dtype=np.float32))
    return mask[None, :, :]  # add batch dimension

mask = causal_mask(Q.shape[1])
out_masked, weights_masked = scaled_dot_product_attention(Q, K, V, mask=mask)
print("Attention Output with Causal Mask:\n", out_masked)
print("Attention Weights with Causal Mask:\n", weights_masked)

```
